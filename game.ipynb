Normalize categorical text

```python
def clean_text(col):
    return col.str.lower().str.strip()

df['Genres'] = clean_text(df['Genres'])
df['Team'] = clean_text(df['Team']) 
df['Title'] = clean_text(df['Title'])
```

```python
df1['Platform'] = clean_text(df1['Platform'])
df1['Genre'] = clean_text(df1['Genre'])
df1['Publisher'] = clean_text(df1['Publisher'])
df1['Name'] = clean_text(df1['Name'])

df1.dropna(subset=['Year'], inplace=True)
df1['Year'] = df1['Year'].astype(int)
```

```python
import sqlite3
```

```python
conn = sqlite3.connect("video_games.db")

cursor = conn.cursor()
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS games_metadata(
id INTEGER PRIMARY KEY AUTOINCREMENT,
title TEXT,
release_date TEXT,
team TEXT,
rating REAL,
times_listed INTEGER,
num_reviews INTEGER,
genres TEXT,
plays REAL,
playing REAL,
backlogs REAL,
wishlist REAL
)
""")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS sales_data(
id INTEGER PRIMARY KEY AUTOINCREMENT,
name TEXT,
platform TEXT,
year INTEGER,
genre TEXT,
publisher TEXT,
NA_sales REAL,
EU_sales REAL,
JP_sales REAL,
Other_sales REAL,
Global_sales REAL
)
""")
```

```python
df_sql = df[[
'Title','Release Date','Team','Rating',
'Times Listed','Number of Reviews','Genres',
'Plays','Playing','Backlogs','Wishlist'
]]

df_sql.columns = [
'title','release_date','team','rating',
'times_listed','num_reviews','genres',
'plays','playing','backlogs','wishlist'
]

# IMPORTANT FIX
df_sql['release_date'] = df_sql['release_date'].astype(str)

df_sql.to_sql("df_metadata", conn, if_exists="append", index=False)
```

```python
df1_sql = df1[[
'Name','Platform','Year','Genre','Publisher',
'NA_Sales','EU_Sales','JP_Sales','Other_Sales','Global_Sales'
]]

df1_sql.columns = [
'name','platform','year','genre','publisher',
'NA_sales','EU_sales','JP_sales','Other_sales','Global_sales'
]

df1_sql.to_sql("sales_data", conn, if_exists="append", index=False)
```

```python
query = """
SELECT
g.title,
g.team,
g.rating,
g.genres,
g.plays,
g.wishlist,
g.backlogs,
s.platform,
s.publisher,
s.genre,
s.NA_sales,
s.EU_sales,
s.JP_sales,
s.Global_sales
FROM games_metadata g
JOIN sales_data s
ON g.title = s.name
"""

merged = pd.read_sql(query, conn)

print(merged.head())
```
